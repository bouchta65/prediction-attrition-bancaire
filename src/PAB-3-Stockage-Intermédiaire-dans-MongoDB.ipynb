# Stockage intermédiaire dans MongoDB

```python
from pyspark.sql import SparkSession
import pandas as pd
from pymongo import MongoClient
spark = SparkSession.builder.getOrCreate()

file_path = "../data/dataset-68f599a4c9b84581895311-6907667a3f0be705109390.csv"

df_preprocessed = spark.read.csv("../data/db_final.csv", header=True, inferSchema=True)
df_preprocessed.show(5)
```

### Conversion du DataFrame Spark en DataFrame Pandas

```python
df_pandas = df_preprocessed.toPandas()
```

### Transformation du DataFrame Pandas en liste de dictionnaires

```python
records = df_pandas.to_dict("records")
```

### Connexion à la base MongoDB et sélection de la collection

```python
client = MongoClient("mongodb://localhost:27017/")
db = client["BankAttritionDB"]             
collection = db["clients_preprocessed"]   
```

### Insertion des données et fermeture de la connexion

```python
if len(records) > 0:
    collection.insert_many(records)
    print("Données prétraitées stockées avec succès dans MongoDB.")
else:
    print("Aucune donnée à insérer.")

client.close()
```
